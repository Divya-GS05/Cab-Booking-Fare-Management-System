```python
# Function 1 - Book a Ride
Rides = []
def book_ride():
  Name = input("Enter Customer Name : ")
  Pickup = input("Enter the pickup location : ")
  Destination = input("Enter the drop location : ")
  Distance = float(input("Enter the distance to travel : "))
  Ride = {"Name" : Name, "Pickup" : Pickup, "Destination" : Destination, "Distance" : Distance}
  Rides.append(Ride)
  print("\nRide Booked:", Ride)
  return Ride

# Function 2 - Calculate Total Fare
def calculate_fare(distance):
  Base_Fare = 50
  if 0 <= distance <= 5:
    Cost = 15
  elif 6 <= distance <= 15:
    Cost = 12
  else:
    Cost = 10

  Total_Fare = distance * Cost + Base_Fare
  return Total_Fare

# Function 3 - Ride History
def ride_history():
  if not Rides:
    print("Ride is not yet Booked")
    return
  for r in Rides:
    print("---------------------------------")
    print(f"Name: {r['Name']}")
    print(f"Pickup: {r['Pickup']}")
    print(f"Destination: {r['Destination']}")
    print(f"Distance: {r['Distance']} KM")
    fare = calculate_fare(r['Distance'])
    print(f"Total fare: {fare}")
    print("---------------------------------")

# Function 4 - Search Ride (FIXED)
def Search_Ride():
  Name = input("Enter Customer Name to Search: ")
  found = False
  for r in Rides:
    if r['Name'].lower() == Name.lower():
      print("---------------------------------")
      print(f"Name: {r['Name']}")
      print(f"Pickup: {r['Pickup']}")
      print(f"Destination: {r['Destination']}")
      print(f"Distance: {r['Distance']} KM")
      fare = calculate_fare(r['Distance'])
      print(f"Total fare: {fare}")
      print("---------------------------------")
      found = True

  if not found:
    print("Ride is not yet Booked")

# Function 5 - Revenue Report
def revenue_report():
  Total_Revenue = sum(calculate_fare(r['Distance']) for r in Rides)
  print(f"Total Revenue: {Total_Revenue}")
  Total_Rides = len(Rides)
  print(f"Total Rides: {Total_Rides}")
  if Total_Rides > 0:
    Average_Fare = Total_Revenue / Total_Rides
    print(f"Average Fare: {Average_Fare}")
  else:
    print("No rides to calculate average fare.")

# bonus 1 - highest distance
def highest_distance():
  if not Rides:
    print("Ride is not yet Booked")
  else:
    print(max(Rides, key=lambda x: x['Distance']))

#bonus 2 - highest fare
def highest_fare():
  if not Rides:
    print("Ride is not yet Booked")
  else:
    print(max(Rides, key=lambda x: calculate_fare(x['Distance'])))


# Display Menu
while True:
  print("\n------CAB BOOKING SYSTEM ------")
  print("\n1. Book Ride")
  print("\n2. View Ride History")
  print("\n3. Search Ride")
  print("\n4. Revenue Report")
  print("\n5.Longest distance")
  print("\n6.Highest fare")
  print("\n7. Exit")
  print("---------------------------------")
  Choice = int(input("Enter your choice: "))
  print("---------------------------------")

  if Choice == 1:
    print("\n------Book a Ride------")
    ride = book_ride()
  elif Choice == 2:
    print("\n------Ride History------")
    ride_history()
  elif Choice == 3:
    print("\n------Search Ride------")
    Search_Ride()
  elif Choice == 4:
    print("\n------Revenue Report------")
    revenue_report()
  elif Choice == 5:
    print("\n------Longest Distance------")
    highest_distance()
  elif Choice == 6:
    print("\n------Highest Fare------")
    highest_fare()
  elif Choice == 7:
    print("Exiting the system. Goodbye!")
    break
```
